## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Using the framing ML problems framework, the details of our Bleed Tracker are structured as follows:
* **The Decision**: Which top-tier pages must an editor triage and refresh first?
* **The Action**: A digital content manager or SEO strategist will execute a structural rewrite or metadata optimization on the flagged URLs.
* **The Cost**: A false positive (recommending a healthy page for triage) wastes valuable editorial hours on normal seasonal variance. A false negative (failing to flag a page in decline) results in a catastrophic missed decline on a core revenue-driving page, causing permanent traffic and revenue loss.
* **The Metric**: Because editorial capacity is tightly capped, a generic metric like ROC-AUC (which evaluates sorting capability across all ranks) is not enough. We declare **Precision@50** as our primary metric to ensure that the top 50 flagged pages are highly likely to be true declines.

```python
print("Primary evaluation metric: Precision@50")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

To establish an honest baseline of active pages, we filter the starter dataset (`data/raw/content_refresh_anonymized.csv`) to include only pages with `impressions_90d > 0` and `content_age_days >= 90` (all 30,000 pages in this initial dataset meet these criteria). The data highlights three key insights that justify our project:

1. **Massive Corporate Asset**: The active pages represent a total of **156,010,989 aggregate search impressions** over the past 90 days. This highlights the substantial traffic volume and revenue potential we are protecting.
2. **Systemic Traffic Erosion**: The base rate of active pages currently tagged with `trend_direction == "down"` is **54.21%** (16,262 out of 30,000 pages). This high percentage proves that traffic decline is a widespread, systemic issue requiring automation.
3. **Behavioral Variance Defeats Simple Rules**: The standard deviation of the `scroll_rate` metric is **29.47%** (relative to a mean of **18.21%**). This extremely high variance indicates that user behavior is noisy and highly diverse across pages, which proves that simple heuristic visibility rules are insufficient and a multivariate ML approach is required.

```python
import pandas as pd
import numpy as np


df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Establish active pages baseline
active_df = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)]

# 1. Total aggregate search impressions
total_impressions = active_df['impressions_90d'].sum()

# 2. Base rate of pages tagged with trend_direction == "down"
base_rate_down = (active_df['trend_direction'] == 'down').mean()

# 3. Standard deviation of scroll_rate
scroll_std = active_df['scroll_rate'].std()
scroll_mean = active_df['scroll_rate'].mean()

print(f"Total Active Pages: {len(active_df):,}")
print(f"Total Aggregate Impressions (90d): {total_impressions:,}")
print(f"Base Rate of 'down' Trend: {base_rate_down:.4%}")
print(f"Scroll Rate Mean: {scroll_mean:.2f}% | Scroll Rate Std Dev: {scroll_std:.2f}%")
```
